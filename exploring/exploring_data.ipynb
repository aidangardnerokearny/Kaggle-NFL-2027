# Exploring Data

Let's take a high level overview of the data we have available to us.

```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
```

## combine_results.csv

```python
df = pd.read_csv("../data/combine_results.csv")
df.head().style.format(precision=2).set_caption("First Rows")
```

```python
print(len(df["combine_position"])) # total number of players in dataset
position_counts = df["combine_position"].value_counts()
print(position_counts)
```

```python
is_te = df["combine_position"] == "TE"
is_db = df["combine_position"] == "DB"
is_wr = df["combine_position"] == "WR"
is_dl = df["combine_position"] == "DL"
is_ol = df["combine_position"] == "OL"
masks = {
    "TE": is_te,
    "DB": is_db,
    "WR": is_wr,
    "DL": is_dl,
    "OL": is_ol,
}
```

```python
fig, ax = plt.subplots(figsize=(8,5))
bars = ax.bar(position_counts.index.astype(str), position_counts.values)
ax.bar_label(bars, padding=3)

ax.set_xlabel("Position")
ax.set_ylabel("Count")
ax.set_title("")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
plt.hist(df["draft_year"], bins=3)
plt.xlabel("Draft Year")
plt.ylabel("Count")
plt.tight_layout();
```

```python
plt.hist(df["combine_height"])
plt.xlabel("Height (in)")
plt.ylabel("Count");
```

```python
bins = np.linspace(df["combine_height"].min(), df["combine_height"].max(), 25)

fig, ax = plt.subplots(figsize=(8,5))
for i, (name, m) in enumerate(masks.items()):
    color = f"C{i}"
    ax.hist(df.loc[m, "combine_height"], bins=bins, histtype="step", label=name, color=color)
    ax.hist(df.loc[m, "combine_height"], bins=bins, histtype="stepfilled", label=name, color=color, alpha=0.5)

ax.set_xlabel("Height (in)")
ax.set_ylabel("Counts")
ax.legend()
plt.show()
```

```python
plt.hist(df["combine_weight"])
plt.xlabel("Weight (lbs)")
plt.ylabel("Count")
```

```python
bins = np.linspace(df["combine_weight"].min(), df["combine_weight"].max(), 25)

fig, ax = plt.subplots(figsize=(8,5))
for i, (name, m) in enumerate(masks.items()):
    color = f"C{i}"
    ax.hist(df.loc[m, "combine_weight"], bins=bins, histtype="step", label=name, color=color)
    ax.hist(df.loc[m, "combine_weight"], bins=bins, histtype="stepfilled", label=name, color=color, alpha=0.5)

ax.set_xlabel("Weight (lbs)")
ax.set_ylabel("Counts")
ax.legend()
plt.show()
```
